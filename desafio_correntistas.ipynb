## 📦 1. Criação dos DataFrames

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Clientes
df_clientes = pd.DataFrame({
    'id_cliente': np.arange(1, 101),
    'nome': ['Cliente_' + str(i) for i in range(1, 101)],
    'idade': np.random.choice([25, 30, 35, 40, np.nan], 100),
    'cidade': np.random.choice(['RJ', 'SP', 'MG', None], 100),
    'status_conta': np.random.choice(['Ativa', 'Inativa', None], 100),
    'score_credito': np.random.choice([300, 500, 700, 900, np.nan], 100),
    'renda_mensal': np.random.choice([2000, 4000, 6000, 8000, np.nan], 100),
    'segmento': np.random.choice(['Varejo', 'Alta Renda', None], 100)
})

# Inserindo duplicados
df_clientes = pd.concat([df_clientes, df_clientes.iloc[:5]])

# Transações
df_transacoes = pd.DataFrame({
    'id_cliente': np.random.randint(1, 101, 200),
    'valor_transacao': np.random.choice([100, 500, 1000, 2000, np.nan], 200),
    'tipo': np.random.choice(['Crédito', 'Débito'], 200),
    'mes': np.random.choice(['Jan', 'Fev', 'Mar'], 200)
})

df_clientes.head()
```

```python
print(df_clientes.info())
```

```python
print(df_transacoes.info())
```

### 🧹 PARTE 3 — Limpeza de dados

**Pergunta 5**  
Remova duplicados  

**Pergunta 6**  
Preencha valores faltantes:
- idade → mediana  
- renda → média  
- cidade → "Não Informado"  
- status → ffill  


```python
#Removendo duplicados

df_clientes_novo = df_clientes.drop_duplicates(keep='first')
display(df_clientes_novo)
```

```python
df_clientes_novo['cidade'] = df_clientes_novo.loc['cidade'].fillna('Não Informado')
display(df_clientes_novo)
```

```python
df_clientes_novo['renda_mensal'] = df_clientes_novo['renda_mensal'].fillna((df_clientes_novo['renda_mensal'].mean()).round(2))
display(df_clientes_novo)
```

```python
df_clientes_novo['idade'] = df_clientes_novo['idade'].fillna(df_clientes_novo['idade'].median())
display(df_clientes_novo)
```

```python
df_clientes_novo['status_conta'] = df_clientes_novo['status_conta'].bfill()
display(df_clientes_novo)
```

```python
print(df_clientes_novo.isna().sum())
```

### 🔁 PARTE 4 — Regras de negócio

**Pergunta 7**  
Crie a coluna `perfil_risco`:
- "Alto Risco" → score < 500  
- "Médio" → 500–700  
- "Baixo Risco" → > 700  


```python
df_clientes_novo['perfil_risco'] = np.where(
    df_clientes_novo['score_credito'] > 700, 
    'Baixo Risco', 
    np.where(df_clientes_novo['score_credito'] > 500,
             'Médio Risco', 
             'Alto Risco'
             )
)
display(df_clientes_novo)
```

```python
df_clientes_novo['perfil_risco'].value_counts(normalize=True)*100
```

### 🔗 PARTE 5 — JOIN

**Pergunta 8**  
Faça um inner join entre clientes e transações, considere o dataframe sem duplicidade

**Pergunta 9**  
Faça um left join, considere o dataframe sem duplicidade


```python
df_unificado = pd.merge(
    df_transacoes,
    df_clientes_novo,
    on='id_cliente',
    how='inner'
)
display(df_unificado)
```

```python
df_unificado_left = pd.merge(
    df_clientes_novo,
    df_transacoes,
    on='id_cliente',
    how='left'
)
display(df_unificado_left)
```

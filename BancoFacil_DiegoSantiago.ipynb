# Análise de Dados do Banco Fácil 🏦

```python
import pandas as pd
import numpy as np
```

## 1. Os Dados

Os dados são referentes a transações de cliente dos Banco Fácil, composta por duas tabelas - uma de clientes e uma de transações.

```python
# Base de transações
transacoes = pd.DataFrame({
    'id_cliente': [1, 2, 1, 3, 2, 1, 3, 2],
    'tipo':       ['Crédito', 'Débito', 'Débito', 'Crédito', 'Crédito', 'Débito', 'Débito', 'Crédito'],
    'categoria':  ['Salário', 'Alimentação', 'Transporte', 'Salário', 'Lazer', 'Alimentação', 'Transporte', 'Salário'],
    'valor':      [3200.0, 150.0, None, 4500.0, 200.0, 80.0, None, 2800.0],
    'canal':      ['App', 'Agência', 'App', 'Internet Banking', 'App', 'Agência', 'App', 'Internet Banking']
})

# Base de cadastro de clientes
clientes = pd.DataFrame({
    'id_cliente': [1, 2, 3, 4],
    'nome':       ['Ana', 'Bruno', 'Carlos', 'Diana'],
    'estado':     ['SP', 'RJ', 'MG', 'SP'],
    'perfil':     ['Premium', 'Básico', 'Intermediário', 'Básico']
})
```

## 2. Conhecendo os Dados

### --- Tabela de Clientes ---

```python
# Tamanho da Tabela
print('O Tamanho da Tabela de clientes é: ', clientes.shape)
print('*'*20)

# Tipos de Dados
print('Os tipos de dados presentes na tabela de clientes são: \n', clientes.dtypes)
print('*'*20)

# Formato da Tabela
display(clientes.head())
print('*'*20)

# Células Vazias
print('Número de células vazias em cada coluna: \n', clientes.isna().sum())

```

### --- Tabela de Transações ---

```python
# Tamanho da Tabela
print('O Tamanho da Tabela de transações é: ', transacoes.shape)
print('*'*20)

# Tipos de Dados
print('Os tipos de dados presentes na tabela de transações são: \n', transacoes.dtypes)
print('*'*20)

# Formato da Tabela
display(transacoes.head(8))
print('*'*20)

# Campos nulos
print('Número de campos nulos em cada coluna: \n', transacoes.isna().sum())
```

### --- Análise ---

```python
print(f'O Banco Fácil possui: {len(clientes)} clientes cadastrados.\n')
print(f'O Banco Fácil registrou: {len(transacoes)} transações no período.\n')
print(f'Existem {transacoes['valor'].isna().sum()} transações sem valor correspondente.\n')
```

## 3. Limpeza dos Dados

```python
# Preenchendo os campos nulos com a mediana dos valores de transação

transacoes['valor'] = transacoes['valor'].fillna(transacoes['valor'].median())
display(transacoes)

# Confirmando se há campos nulos
print('Número de campos nulos em cada coluna: \n', transacoes.isna().sum())
print(f'\nExistem {transacoes['valor'].isna().sum()} transações sem valor correspondente.\n')
```

## 4. Cálculos sobre valores

```python
# Criando array numpy da coluna valor
transacoes_valores = np.array(transacoes['valor'])
print(transacoes_valores)
print(type(transacoes_valores))
```

```python
# Total de Transações
print(f'O Valor total de transações no período foi de: R$ {transacoes_valores.sum()}\n')
# Valor Médio das Transações
print(f'O Valor médio das transações no período foi de: R$ {transacoes_valores.mean()}\n')
# Valor mínimo de Transação
print(f'O Valor mínimo das transações no período foi de: R$ {transacoes_valores.min()}\n')
# Valor Máximo de Transação
print(f'O Valor máximo das transações no período foi de: R$ {transacoes_valores.max()}\n')
```

## 5. Relacionamento entre as Bases

```python
# Fazendo um Join com inner
df_inner = pd.merge(
    transacoes,
    clientes,
    on='id_cliente',
    how='inner'
)
display(df_inner)

# Fazendo um Join com left partindo da tabela clientes

df_left = pd.merge(
    clientes,
    transacoes,
    on='id_cliente',
    how='left'
)
display(df_left)
```

```python
# Comparando o shape dos dois DataFrames
print(
    f'O DataFrame feito pelo inner partindo das transações tem: {df_inner.shape[0]} linhas e {df_inner.shape[1]} colunas\n'
    f'Enquanto o DataFrame feito pelo left partindo da tabela clientes tem: {df_left.shape[0]} linhas e {df_left.shape[1]} colunas'
)
```

```python
# Investigando os campos nulos do df_left

display(df_left.loc[df_left['valor'].isna()])
```

### --- Análise ---

O DataFrame gerado pelo `pd.merge()` a partir da tabela clientes possui uma linha a mais.

A cliente **Diana** que encontra-se cadastrada na tabela de clientes, não realizou nenhuma transação no período. 

Neste modo `left` do `pd.merge()` o pandas relaciona as transações com os clientes, preenchendo com campos nulos clientes que não realizaram transações no período.

## 6. Analise do Perfil

```python
# Analisando o perfil pela soma e média de transações dos clientes

perfil = df_inner.groupby('perfil').agg(
    soma   =   ('valor', 'sum'),
    media   =   ('valor', 'mean')
).round(2).reset_index()

display(perfil)
```

### --- Análise ---

```python
# Perfil com maior média de transações
print('O Perfil que teve a maior média de transações foi: ', 
      perfil.loc[perfil['media'] == perfil['media'].max(), 'perfil'].values[0],
      ', com R$',
      perfil['media'].max().round(2),
      'transações.'
      )
```
